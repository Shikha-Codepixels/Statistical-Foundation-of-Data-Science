```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Student': ['Mahi', 'Ritvik', 'Mira', 'Sumit', 'Manvi'],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female'],
    'Department': ['CS', 'DS', 'CS', 'DS', 'CS'],
    'Math': [85, 72, 91, 68, 88],
    'Physics': [78, 81, 89, 74, 93],
    'Chemistry': [92, 88, 95, 79, 90]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("First 5 rows: ")
display(df.head());
print("\nDataset Shape: ", df.shape)
```

```python
marks = df[['Math', 'Physics', 'Chemistry']].to_numpy()
print("Numpy Array: ")
print(marks)
```

```python
print("Shape: ", marks.shape)
print("Size: ",  marks.size)
print("Dimensions: ", marks.ndim)
```

```python
additional_marks = np.array([
    [5, 4, 3],
    [4, 5, 2],
    [3, 4, 5],
    [5, 3, 4],
    [2, 5, 3]
])
addition = marks + additional_marks
print("Addition: ")
print(addition)

subtraction = marks - additional_marks
print("\nSubtraction: ")
print(subtraction)

multiplication = marks * additional_marks
print("\nMultiplication: ")
print(multiplication)

transpose = marks.T
print("\nOriginal Matrix: ")
print(marks)
print("\nTranspose:")
print(transpose)

matrix_product = marks.T @ marks
print("\nMatrix Product: ")
print(matrix_product)

inverse = np.linalg.inv(matrix_product)
print("\nInverse of the matrix: ")
print(inverse)
```

```python
encoded_df = df.copy()
encoded_df['Gender_Encoded'] = encoded_df['Gender'].map({'Female': 1, 'Male': 0})
encoded_df['Department_Encoded'] = encoded_df['Department'].map({'CS': 0, 'DS': 1})
print(encoded_df)
```

```python
print("Final Dataset: ")
print(encoded_df)
```
