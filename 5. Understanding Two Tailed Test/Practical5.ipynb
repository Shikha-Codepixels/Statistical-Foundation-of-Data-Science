# Probability of evaluation score > 4.5

```python
mean = df['eval'].mean()
std = df['eval'].std()
x = 4.5
z = (x - mean) / std
# probability P(X > 4.5)
probability = 1 - norm.cdf(z)
print("Mean =", mean)
print("Standard Deviation =", std)
print("Z-score =", z)
print("Probability of evaluation score > 4.5 =", probability)
print("Probability (%) =", probability * 100)
```

**The probability of receiving an evaluation score greater than 4.5 is approximately 18.29%.**

# Probability of evaluation score > 3.5 and < 4.2
P(3.5 < X < 4.2)

```python
lower = 3.5
upper = 4.2
z_lower = (lower - mean) / std
z_upper = (upper - mean) / std
probability_range = norm.cdf(z_upper) - norm.cdf(z_lower)
print("Z-score lower =", z_lower)
print("Z-score upper =", z_upper)
print("Probability =", probability_range)
print("Probability (%) =", probability_range * 100)
```

**The probability of receiving an evaluation score between 3.5 and 4.2 is approximately 45%.**

# Two tailed z-test

```python
# Given values :
mu = 12
sigma = 5.5
sample_mean = 10.7
n = 36
alpha = 0.05

standard_error = sigma / np.sqrt(n)
z_stat = (sample_mean - mu) / standard_error
p_value = 2 * (1 - norm.cdf(abs(z_stat)))
print("Standard error =", standard_error)
print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

Null Hypothesis
$$ H_0: \mu = 12 $$

The mean score of the regional players is not significantly different from the historical mean of 12 points.

Alternative Hypothesis
$$ H_1: \mu \neq 12 $$

The mean score of the regional players is significantly different from the historical mean of 12 points.

```python
"critical_value = norm.ppf(1 - alpha / 2)
print("Critical values:", -critical_value, "and", critical_value)

if abs(z_stat) > critical_value:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

The calculated Z-statistic is -1.42, while the critical values at a 5% significance level for a two-tailed test are -1.96 and +1.96. Since the calculated Z-statistic lies within the acceptance region (-1.96, 1.96), we fail to reject the null hypothesis
